```python
### work flow
from typing_extensions import TypedDict
from langchain_core.messages import AnyMessage  #ut can ai message or human message
from typing import Annotated   #labling
from langgraph.graph.message import add_messages  # reducers in langgraph
```

```python
class State(TypedDict):


    # list[messgae]  like append the messgae one by one,  add_messages(reducers not ovride it will help to append)
    messages=Annotated[list[AnyMessage], add_messages]
```

```python
#### create the bot
from IPython.display import Image, display
from langgraph.graph import StateGraph, START, END
from langgraph.prebuilt import ToolNode
from  langgraph.prebuilt import tools_condition  # if my llm call tool what condition i need to call
```

```python
def tool_calling_llm(state:State):
    return{"message":[llm_with_tools.invoke(State["messages"])]}
```

```python
#biuld the graph
```

```python
buildr_graph1=StateGraph(State)
buildr_graph1.add_node("tool_calling_llm", tool_calling_llm)
buildr_graph1.add_node("tools", ToolNode(tools))
```

```python
### check the working
```

```python
### ad the edges to connect


buildr_graph1.add_edge(START, "tool_calling_llm")
buildr_graph1.add_conditional_edges(
    "tool_calling_llm",
    lambda state: "to_tools" if state.get("last_message", {}).get("tool_call") else "to_end",
    {
        "to_tools": "tools",
        "to_end": END,
    }
)

buildr_graph1.add_edge("tools", END)
graph=buildr_graph1.compile()
```

first chat bot

```python
from langchain.schema import HumanMessage

def chat_with_graph(user_input: str):
    # Wrap user input in a list of HumanMessage
    input_messages = [HumanMessage(content=user_input)]

    # Invoke the graph (stream_mode is optional; can use invoke for final output)
    output = graph.invoke(input_messages)

    # Print the output safely
    if isinstance(output, list):
        for msg in output:
            # If msg is a BaseMessage object
            if hasattr(msg, "content"):
                print("Assistant:", msg.content)
            else:
                print("Assistant:", msg)
    else:
        # For string or single object
        print("Assistant:", output)


# Interactive loop
while True:
    try:
        user_input = input("User: ")
        if user_input.lower() in ["quit", "exit", "q"]:
            print("Goodbye!")
            break
        chat_with_graph(user_input)
    except EOFError:
        user_input = "Hello, tell me about your tools."
        print("User:", user_input)
        chat_with_graph(user_input)
        break
```
