```python
## when ever my graph is exicuted it return in form schema dict 
### state
### first deine the state of the graph , the state schema srever as the input schema for all the nodes and edges in the graph

from pydantic import BaseModel
from typing_extensions import TypedDict
class State(TypedDict):
    graph_info: str
```

```python
# node
def state_play(State):
    print("start play node is called")
    return {"graph_info":State["graph_info"]+"i am planning to play"}


def circket(State):
    print("start play node is called")
    return {"graph_info":State["graph_info"]+"i am planning to play cricket"}

def basketball(State):
    print("start play node is called")
    return {"graph_info":State["graph_info"]+"i am planning to play the basketball"}



```

```python
# to decide where to choice

import random
from typing import Literal
def random_return_whichone(State)-> Literal['circket', 'basketball']:
    if random.random()>0.5:
        return "circket"
    else:
        return "basketball"
```

```python
from IPython.display import Image,display

# this is will disply the iamge architecture
from langgraph.graph import StateGraph, START, END

# add the all the nodes
graph = StateGraph(State)
graph.add_node("state_play", state_play)
graph.add_node("cricket", circket)       # ✅ spelling fixed
graph.add_node("basketball", basketball)

graph.add_edge(START, "state_play")
graph.add_conditional_edges("state_play", random_return_whichone)
graph.add_edge("cricket", END)
graph.add_edge("basketball", END)

# --- compile & display ---
graph_builder = graph.compile()
display(Image(graph_builder.get_graph().draw_mermaid_png()))
```
